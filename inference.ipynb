# Inference Notebook

## Import Libraries

```python
import joblib
import numpy as np
import pandas as pd
```

## Load Model

```python
model = joblib.load("best_model.pkl")
```

## Load sample data

```python
sample = pd.DataFrame([{
    "grid_position" : 1.0, # Pole position
    "quali_position" : 1.0, # Qualified first
    "round" : 8, # 8th race of the season
    "year" : 2024,
    "driver_points_before" : 195.0, # Comfortably leading the championship
    "driver_position_before" : 1.0, # Championship leader
    "driver_wins_before" : 5.0, # 5 wins this season already
    "driver_age" : 26.5,
    "driver_nationality" : "Dutch",
    "constructor_nationality" : "British",
    "constructor_position_before" : 1.0, # Constructor also leading
    "driver_historical_dnf_rate" : 0.07, # Very reliable — 7% DNF rate
    "pit_stops" : 2.0}]) # Standard 2-stop race
```

## Predict sample data

```python
prediction  = model.predict(sample)[0]
probability = model.predict_proba(sample)[0]

print(f"Prediction: {'Podium' if prediction == 1 else 'Not Podium'}")
print(f"Probability: {probability[1]:%}")
```
